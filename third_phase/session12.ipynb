```python
def my_func(fun):
    def wrapper():
        print('*****************')
        fun()
        print('*****************')
    return wrapper
    
def chal():
    print('Pagal hai kya')

a = my_func(chal)
a()
```

```python
def my_func(fun):
    def wrapper():
        print('*****************')
        fun()
        print('*****************')
    return wrapper

@my_func
def chal():
    print('Pagal hai kya')

chal()
```

```python
import time
def timer(k):
    a = time.time()
    for i in range(k):
        print('Deepak Singh')
    b = time.time()
    print(b-a)
timer(6)    
```

```python
import time

def timer(func):
    def wrapper(*args):
        start = time.time()
        func(*args)
        end = time.time()
        print('The time to execute this  function is : ',end-start)
    return wrapper


@timer
def tum():
    print('Tu mera chela hai')
tum()

@timer
def Pagal():
    print('Kinjal Khuti.... hai')
Pagal()    
```

```python
def sanity_check(data_type):
  def outer_wrapper(func):
    def inner_wrapper(*args):
      if type(*args) == data_type:
        func(*args)
      else:
        raise TypeError('Ye datatype nai chalega')
    return inner_wrapper
  return outer_wrapper

@sanity_check(int)
def square(num):
  print(num**2)

@sanity_check(str)
def greet(name):
  print('hello',name)

square(2)
```
