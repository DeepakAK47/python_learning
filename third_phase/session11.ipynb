```python
# IMP
class Bank:
    def __init__(self,balance):
        self.balance = balance
    def withdraw(self,amt):
        if amt<0:
            raise Exception('Amount can not be negative')
        if amt > self.balance:
            raise Exception('Saake gareeb tere account mei balance nhi hai')
        else:
            self.balance = self.balance - amt

obj = Bank(1000)
try:
    obj.withdraw(50000)
except Exception as e:
    print(e)
else:
    print(obj.balance)
    
```
