```python
with open('sample.txt','r') as f:
    chunk_size = 100
    while len(f.read(chunk_size)) > 0 :
        s = f.read(100)
        print(s)
        f.read(chunk_size)
    
```

```python
with open('sample.txt','r') as f:
    print(f.read(10))
    print(f.tell()) # f.tell() tell the current cursor position
    print(f.read(10))
    f.seek(0)
    print(f.read(10))
    print(f.tell()) # It is used for adjusting the cursor position
```

# Serialization and Deserialization

```python
import json
L = [1,2,3,4,5]
with open('sample.json','w') as f:
    json.dump(L,f)
```

```python
d = {
    'name':'Deepak Singh',
    'class':'AIDS',
    'Enrollment Number' : 49,
    'Subject':'DBMS'
}
with open('sample2.json','w') as f:
    json.dump(d,f,indent=4)
```

```python
import json
with open('sample2.json','r') as f:
    d = json.load(f)
    print(type(d))
    print(d)
```

```python
class Person:
    def __init__(self,name,age):
        self.name = name
        self.age = age
    def info(self):
        print(f"My name is {self.name} and my age is {self.age}")
```

```python
P = Person('Deepak Signh',22)
```

```python
# Picle load
import pickle
with open('sample3.json','wb') as f:
    pickle.dump(P,f)
```

```python
# picle dump
import pickle 
with open('sample3.json','rb') as f:
    p = pickle.load(f)
P.info()    
```
